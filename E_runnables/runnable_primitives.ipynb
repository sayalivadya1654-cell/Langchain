```python
# runnablesequence
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# LLM
llm = ChatOllama(model="llama3.2")

# Prompt
prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in one simple sentence."
)

# Parser
parser = StrOutputParser()

# Runnable Sequence
chain = prompt | llm | parser

# Run
result = chain.invoke({
    "topic": "Machine Learning"
})

print(result)
```

```python
# runnableparallel
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2")

# First chain
prompt1 = ChatPromptTemplate.from_template(
    "Explain {topic} in one simple sentence."
)

chain1 = prompt1 | llm | StrOutputParser()

# Second chain
prompt2 = ChatPromptTemplate.from_template(
    "Give one simple example of {topic}."
)

chain2 = prompt2 | llm | StrOutputParser()

# Run both in parallel
parallel_chain = RunnableParallel(
    explanation=chain1,
    example=chain2
)

result = parallel_chain.invoke({
    "topic": "Machine Learning"
})

print("Explanation:", result["explanation"])
print("Example:", result["example"])
```

```python
# runnablelambda
from langchain_ollama import ChatOllama
from langchain_core.runnables import RunnableLambda
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2")

# Python function
def make_question(topic):
    return f"Explain {topic} in simple words."

# Convert function into Runnable
question = RunnableLambda(make_question)

# Connect Runnables
chain = question | llm | StrOutputParser()

result = chain.invoke("Deep Learning")

print(result)
```

```python
# runnablebranch
from langchain_ollama import ChatOllama
from langchain_core.runnables import RunnableBranch, RunnableLambda
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model="llama3.2")

python_prompt = RunnableLambda(
    lambda x: f"Explain {x} as a Python topic."
)

general_prompt = RunnableLambda(
    lambda x: f"Explain {x} in simple words."
)

python_chain = python_prompt | llm | StrOutputParser()

general_chain = general_prompt | llm | StrOutputParser()

branch = RunnableBranch(
    (
        lambda x: "python" in x.lower(),
        python_chain
    ),
    general_chain
)

result = branch.invoke("Python")

print(result)
```
