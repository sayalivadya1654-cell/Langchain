```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel

# Model
llm = ChatOllama(model="llama3.2")

# Prompts
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize this text in one sentence: {text}"
)

feature_prompt = ChatPromptTemplate.from_template(
    "Extract the main features from this text: {text}"
)

# Chains
summary_chain = summary_prompt | llm
feature_chain = feature_prompt | llm

# Parallel Chain
parallel_chain = RunnableParallel(
    summary=summary_chain,
    features=feature_chain
)

# Input
result = parallel_chain.invoke({
    "text": "This phone has a 5000mAh battery and a powerful camera."
})

# Output
print("SUMMARY:")
print(result["summary"].content)

print("\nFEATURES:")
print(result["features"].content)
```
