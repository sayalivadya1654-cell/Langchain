```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Create Ollama model
llm = ChatOllama(model="llama3.2:latest")

# 2. First prompt
prompt1 = ChatPromptTemplate.from_template(
    "Explain {topic} in simple words."
)

# 3. Second prompt
prompt2 = ChatPromptTemplate.from_template(
    "Summarize the following explanation in 3 bullet points:\n\n{explanation}"
)

# 4. Output parser
parser = StrOutputParser()

# 5. First chain
chain1 = prompt1 | llm | parser

# 6. Second chain
chain2 = prompt2 | llm | parser

# 7. Connect both chains sequentially
sequential_chain = chain1 | chain2

# 8. Run the chain
result = sequential_chain.invoke({
    "topic": "Machine Learning"
})

print(result) 

sequential_chain.get_graph().print_ascii() 
```
