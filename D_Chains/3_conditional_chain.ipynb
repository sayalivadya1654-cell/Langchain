```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableBranch

# LLM
llm = ChatOllama(model="llama3.2")

# Python prompt
python_prompt = ChatPromptTemplate.from_template(
    "You are a Python expert. Answer this question simply: {question}"
)

# SQL prompt
sql_prompt = ChatPromptTemplate.from_template(
    "You are a SQL expert. Answer this question simply: {question}"
)

# Chains
python_chain = python_prompt | llm
sql_chain = sql_prompt | llm

# Conditional Chain
conditional_chain = RunnableBranch(
    (
        lambda x: "python" in x["question"].lower(),
        python_chain
    ),
    (
        lambda x: "sql" in x["question"].lower(),
        sql_chain
    ),
    llm
)

# Input
result = conditional_chain.invoke({
    "question": "What is a Python list?"
})

print(result.content)
```
