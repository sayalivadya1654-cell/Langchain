```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

# chat template 
chat_template = ChatPromptTemplate([
    ('system','you are a helpful customer support agent'),
    MessagesPlaceholder(variable_name='chat_history'),
    ('human','{query}')
])

chat_history = []

# load chat history 
with open('chat_history.txt') as f: 
    chat_history.extend(f.readlines())
print(chat_history)

# prompt 
prompt = chat_template.invoke({'chat_history':chat_history,'query':"Can I change the delivery address?"})

print(prompt)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

# chat template 
chat_template = ChatPromptTemplate([
    ('system', 'you are a helpful customer support agent'),
    MessagesPlaceholder(variable_name='chat_history'),
    ('human', '{query}')
])

chat_history = []

# load chat history 
with open('chat_history.txt') as f:
    for line in f:
        line = line.strip()

        if line.startswith('Human:'):
            chat_history.append(HumanMessage(content=line.replace('Human:', '').strip()))

        elif line.startswith('AI:'):
            chat_history.append(AIMessage(content=line.replace('AI:', '').strip()))

print(chat_history)

# prompt 
prompt = chat_template.invoke({
    'chat_history': chat_history,
    'query': 'Can I change the delivery address?'
})

print(prompt)
```
